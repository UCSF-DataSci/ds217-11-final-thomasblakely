## Setup

````python
# Import libraries
import pandas as pd
import numpy as np
import os

# Load cleaned data from Q2
df = pd.read_csv('output/q2_cleaned_data.csv')
print(f"Loaded {len(df):,} cleaned records")
````

## Preparing Data for Temporal Feature Extraction

````python
# 1. `output/q3_wrangled_data.csv`

# Begin Chunk
print('Setting-up Cleaned Dataframe for Temporal Feature Extraction')

# Convert to Datetime
df['Measurement Timestamp'] = pd.to_datetime(df['Measurement Timestamp'])
df['Measurement Timestamp Label'] = pd.to_datetime(df['Measurement Timestamp Label'])

# Set Index
df = df.set_index('Measurement Timestamp')

# Sort Index Chronologically
df = df.sort_index()

# Output q3_wrangled_data.csv
df.reset_index().to_csv('output/q3_wrangled_data.csv', index = False)

# End Chunk
print('Temporal set-up successful and CSV saved at: output/q3_wrangled_data.csv')
````

## Generating Temporal Features

````python
# 2. `output/q3_temporal_features.csv`

# Begin chunk
print('Creating Measurement Dictionary ...')

# Measurement Dictionary
measurements = {
    'Measurement Timestamp' : df.index,
    'hour' : df.index.hour,
    'day_of_week' : df.index.weekday,
    'month' : df.index.month,
    'year' : df.index.year,
    'day_name' : df.index.day_name(),
    'is_weekend' : (df.index.weekday > 4).astype(int)
}

# Convert Dictionary to Dataframe
measurements_df = pd.DataFrame(measurements)
print(f'First row of dictionary:\n{measurements_df.head(1)}')

# Output q3_temporal_features.csv
measurements_df.to_csv('output/q3_temporal_features.csv', index = False)

# End chunk
print('Measurement Dictionary has successfully been created and stored at: output/q3_temporal_features.csv')
````

## Outputting Important Datetime Info

````python
# 3. `output/q3_datetime_info.txt`

# Begin chunk
print('Creating a Datetime Info text file ...')

with open('output/q3_datetime_info.txt', 'w') as f:
    f.write('Date Range After Datetime Parsing:\n')
    f.write(f'Start: {df.index.min()}\n')
    f.write(f'End: {df.index.max()}\n')
    f.write(f'Total Duration: {df.index.max() - df.index.min()}')

# End chunk
print('Datetime Info file has successfully been created and stored at: output/q3_datetime_info.txt')
````
